```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv('../data/sales_cleaned.csv',parse_dates=['Order_Date'])
df.head()
```

```python
print(df.info())
print('Duplicate rows:',df.duplicated().sum())
print('Missing values:\n',df.isna().sum())
```

```python
total_sales=df.Sales.sum(); total_profit=df.Profit.sum(); orders=df.Order_ID.nunique(); aov=total_sales/orders
print({'Total Sales':total_sales,'Total Profit':total_profit,'Orders':orders,'AOV':aov})
```

```python
print(df.groupby('Category').agg(Sales=('Sales','sum'),Profit=('Profit','sum')).sort_values('Sales',ascending=False))
print(df.groupby('Region').Sales.sum().sort_values(ascending=False))
print(df.groupby('Product').Sales.sum().sort_values(ascending=False).head(10))
```

```python
monthly=df.groupby(df.Order_Date.dt.to_period('M')).Sales.sum()
monthly.plot(figsize=(10,4),marker='o',title='Monthly Sales Trend'); plt.tight_layout()
```
